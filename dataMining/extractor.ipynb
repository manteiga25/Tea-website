```python
import pandas as pd
import numpy as np
import sqlite3
```

```python
df = pd.read_json("teas.json")
df
```

```python
print(df.columns)
```

```python
def preprocessDescription(text :str):
    return text.replace("NEW! ", "").replace(" Get $4 off your 6-pack with code EXTRA4.** **This offer is not combinable with other discounts", "").replace("*", "")
```

```python
df["description"] = df["description"].apply(preprocessDescription)
df["description"]
```

```python
def preprocessCaffeine(text: str):
    return text == "no"
```

```python
df["caffeine_free"] = df["caffeine_free"].apply(preprocessCaffeine)
df["caffeine_free"]
```

```python
df["how_to_enjoy"] = df["how_to_enjoy"].apply(lambda x: "\n".join(x))
df["how_to_enjoy"][0]
```

```python
db = sqlite3.connect("tea.db")
cursor = db.cursor()
```

```python
dados = (
    df.assign(has_caffeine=df["caffeine_free"])
      .rename(columns={
          "is_true_tea": "is_tea",
          "product_url": "url",
          "image_url": "img_url",
          "how_to_enjoy": "cook",
          "label_warnings": "warning",
      })
)

cursor.executemany("""
    INSERT INTO Tea (
        name,
        description,
        is_tea,
        has_caffeine,
        url,
        img_url,
        price,
        warning,
        cook
    )
    VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
""", dados[
    [
        "name",
        "description",
        "is_tea",
        "has_caffeine",
        "url",
        "img_url",
        "price",
        "warning",
        "cook",
    ]
].values.tolist())
```

```python
data = cursor.execute("SELECT * FROM Tea;").fetchall()
data
```

```python
db.commit()
```

```python
herbs_df = pd.read_csv("herbs.csv")
herbs_df.columns
```
